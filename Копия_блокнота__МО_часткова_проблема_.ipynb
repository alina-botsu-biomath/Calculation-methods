```python
import numpy as np
import math
from scipy.linalg import solve

def lu_solve(A,b):
  n=len(b)
  L=np.eye((n))
  U=A.copy()

  #1. Прямий хід: LU-розклад
  for k in range(n):
    if U[k][k]==0:
      print("Error")
    else:
      for i in range(k+1,n):
        N=U[i][k]/U[k][k]
        L[i][k]=-N
        for j in range(k,n):
          U[i][j]=U[i][j]+N*U[k][j]
  #2 Пряма підстановка: Вирішуємо Ly=b

  y=np.zeros((n,1))
  for k in range(n):
    y[k]=b[k]
    for i in range(k):
      y[k]=y[k]-L[k][i]*y[i]
    y[k]=y[k]/L[k][k]

  #3 Зворотня підстановка: Ux=y

  x=np.zeros((n,1))

  for k in range(n-1,-1,-1):
    x[k]=y[k]
    for i in range(k+1,n):
      x[k]=x[k]-U[k][i]*x[i]
    x[k]=x[k]/U[k][k]

  return x,L,U

A=np.random.rand(4,4)
#A=np.array([[10,2,1],[2,8,1],[1,1,5]],dtype=float)
print("Matrix A:\n", A)
b=np.random.rand(4)
#b=np.array([[2],[5],[8]])
print("vector b:\n", b)

x=lu_solve(A,b)
print("Solution:\n",x)
print("Число обумовленості:\n", np.linalg.cond(A,2))
print("Check A and b:\n", np.dot(A,b))

print(np.linalg.eigvals(A),"власні значення")


```

```python
import numpy as np

n=20
k=2
alp=10+k/10
eps=1e-10
max_i=1000
max_i

A = np.zeros((n, n))

for i in range(n):
    for j in range(n):
        if i==j:
            A[i,j]=alp if i%2==0 else-alp
        else:
            A[i,j]=(i+1+j+1)/alp

print("Матриця A:\n",A)

# cтепеневий метод

def power_method(A, eps=1e-10, max_i=1000):
    n=A.shape[0]
    y0=np.ones(n)
    x=y0/np.linalg.norm(y0)
    l_old=0

    for k in range(max_i):
        # y(k+1) = A*x(k)
        y=A@x
        #lambda(k+1)
        l_new=np.dot(y, x)
        # нормування
        x=y/np.linalg.norm(y)
        # перевірка збіжності
        if abs(l_new-l_old)<=eps*abs(l_new):
            break
        l_old=l_new

    return l_new, x, k + 1

# зворотна ітерація

def inverse_iteration(A, eps=1e-10, max_i=1000):
    n=A.shape[0]
    y0=np.ones(n)
    x=y0/np.linalg.norm(y0)
    l_old=0

    for k in range(max_i):
        # A*y = x
        y=np.linalg.solve(A,x)
        # оцінка власного значення
        mu=np.dot(y,x)
        l_new=1/mu
        # нормування
        x=y/np.linalg.norm(y)
        # критерій зупинки
        if abs(l_new-l_old)<=eps*abs(l_new):
            break
        l_old=l_new
    return l_new, x, k + 1


# зворотна ітерація зі зсувом

def shifted_inverse_iteration(A, sigma, eps=1e-10, max_i=1000):
    n=A.shape[0]
    y0=np.ones(n)
    x=y0/np.linalg.norm(y0)
    I=np.eye(n)
    l_old=0

    for k in range(max_i):
        # (A - sigma*I)y = x
        y=np.linalg.solve(A-sigma * I,x)
        x=y/np.linalg.norm(y)
        # відношення Релея
        l_new=x@A@x
        if abs(l_new-l_old)<=eps*abs(l_new):
            break
        l_old=l_new
    return l_new, x, k + 1

# зворотна ітерація зі зсувом Релея

def rayleigh_iteration(A, eps=1e-10, max_i=1000):
    n=A.shape[0]
    y0=np.ones(n)
    x=y0/np.linalg.norm(y0)
    I=np.eye(n)
    s=x@A@x

    for k in range(max_i):
        # (A - s*I)y = x
        y=np.linalg.solve(A-s*I,x)
        # нормування
        x=y/np.linalg.norm(y)
        # новий зсув Релея
        s_new=x@A@x
        # критерій зупинки
        if abs(s_new-s)<=eps*abs(s_new):
            break
        s=s_new
    return s_new, x, k + 1

print()
#1
# Найбільше власне значення
l_max, vec_max, iter_max=power_method(A,eps)

print("Найбільше власне значення:",l_max)
print("Кількість ітерацій:",iter_max)

# Найменше власне значення
l_min, vec_min, iter_min=inverse_iteration(A,eps)

print("Найменше власне значення:",l_min)
print("Кількість ітерацій:",iter_min)

#5
# Найближче до заданого alpha
sigma = 5
lambda_shift, vec_shift, iter_shift=shifted_inverse_iteration(A,sigma,eps)

print("Власне значення, найближче до sigma =", sigma)
print(lambda_shift)

print("Кількість ітерацій:",iter_shift)

#6
# Порівняння швидкості збіжності

lambda_rayleigh, vec_rayleigh, iter_rayleigh=rayleigh_iteration(A,eps)

print("\nМетод Релея:",lambda_rayleigh)
print("Кількість ітерацій:",iter_rayleigh)

# Перевірка через NumPy
print("Перевірка:\n",np.linalg.eigvals(A))
```
