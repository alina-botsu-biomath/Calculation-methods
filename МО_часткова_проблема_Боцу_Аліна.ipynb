```python
import numpy as np
import math
from scipy.linalg import solve

n=20
k=2
alp=10+k/10
eps=1e-10
max_i=1000
max_i
A = np.zeros((n, n))

#Матриця А
for i in range(n):
    for j in range(n):
        if i==j:
            A[i,j]=alp if i%2==0 else-alp
        else:
            A[i,j]=(i+1+j+1)/alp
np.set_printoptions(precision=2, suppress=True)
print("матриця A:\n",A)

# cтепеневий метод
def pow_met(A, eps=1e-10, max_i=1000):
    n=A.shape[0]
    y0=np.ones(n)
    x=y0/np.linalg.norm(y0)
    l_old=0

    for k in range(max_i):
        y=A@x
        l_new=np.dot(y, x)
        x=y/np.linalg.norm(y)
        if abs(l_new-l_old)<=eps*abs(l_new):
            break
        l_old=l_new
    return l_new, x, k + 1

# зворотна ітерація
def inv_it(A, eps=1e-10, max_i=1000):
    n=A.shape[0]
    y0=np.ones(n)
    x=y0/np.linalg.norm(y0)
    l_old=0

    for k in range(max_i):
        y=np.linalg.solve(A,x)
        mu=np.dot(y,x)
        l_new=1/mu
        x=y/np.linalg.norm(y)
        if abs(l_new-l_old)<=eps*abs(l_new):
            break
        l_old=l_new
    return l_new, x, k + 1

# зворотна ітерація зі зсувом
def shifted_inv_it(A, sigma, eps=1e-10, max_i=1000):
    n=A.shape[0]
    y0=np.ones(n)
    x=y0/np.linalg.norm(y0)
    I=np.eye(n)
    l_old=0

    for k in range(max_i):
        y=np.linalg.solve(A-sigma * I,x)
        x=y/np.linalg.norm(y)
        l_new=x@A@x
        if abs(l_new-l_old)<=eps*abs(l_new):
            break
        l_old=l_new
    return l_new, x, k + 1

# зворотна ітерація зі зсувом Релея
def rayleigh_it(A, eps=1e-10, max_i=1000):
    n=A.shape[0]
    y0=np.ones(n)
    x=y0/np.linalg.norm(y0)
    I=np.eye(n)
    s=x@A@x

    for k in range(max_i):
        y=np.linalg.solve(A-s*I,x)
        x=y/np.linalg.norm(y)
        s_new=x@A@x
        if abs(s_new-s)<=eps*abs(s_new):
            break
        s=s_new
    return s_new, x, k + 1

print()
#1 Найбільше власне значення
l_max, vec_max, iter_max=pow_met(A,eps)
print("найбільше власне значення:",l_max)
print("ітерацій:",iter_max)

# Найменше власне значення
l_min, vec_min, iter_min=inv_it(A,eps)
print("найменше власне значення:",l_min)
print("ітерацій:",iter_min)

#5 Найближче до заданого alpha
lambda_shift, vec_shift, iter_shift=shifted_inv_it(A,alp,eps)
print("власне значення, найближче до alp =", alp)
print("найближче до заданого alpha",lambda_shift)
print("ітерацій:",iter_shift)

#6 Порівняння швидкості збіжності
lambda_rayleigh, vec_rayleigh, iter_rayleigh=rayleigh_it(A,eps)
print("метод Релея:",lambda_rayleigh)
print("ітерацій:",iter_rayleigh)

# Перевірка
print("перевірка:\n",np.linalg.eigvals(A))
```
