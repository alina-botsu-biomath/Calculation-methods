Для симетричної додатньо визначеної матриці А побудувати модельний приклад:
* побудувати довільний вектор x

* обчислити вектор b правої частини

* для матриці A, обчислити її визначник та число обумовленості

Розв'язати СЛАР Ax=b методами:

* Простої ітерації з оптимальним параметром,

* методом Зейделя з коефіцієнтом 0.2,

* методом Річардсона p=5.

Всі розв'язки знаходити з точністю е = 10^-3.

Ці методи реалізуються самостійно.

Можна використовувати методи: numpy.linalg.cond, numpy.linalg.det, numpy.linalg.eigvals, numpy.linalg.norm, A.dot(b).
Неможна використовувати методи, які автоматично розраховують вказані в завданні ітераційні методи.

```python
import numpy as np

#n=3
#A=np.random.rand(n,n)
#b=np.random.rand(n)

A=np.array([[10,2,1],[2,8,1],[1,1,5]],dtype=float)
xT=np.array([[2],[5],[8]])
b=np.dot(A,xT)

n=len(b)

x=np.zeros((n,1))
y=np.zeros((n,1))

print('A \n',A)
print('xT \n',xT)
print('b \n',b)
print(np.linalg.cond(A, 2), "число обумовленості програмно")
print(np.linalg.det(A), "визначник матриці А")

print()
print(np.linalg.eigvals(A),"власні значення")
eigvals = np.linalg.eigvals(A)

m = np.min(eigvals)
M = np.max(eigvals)
print('\nm =', m)
print('M =', M)

print()
alp=2/(m+M)
print(alp,'оптимальний параметр alpha')

n=3
x=np.zeros((n,1))
k=0
eps=1e-3
while True:
    x_new=x-alp*(A.dot(x)-b)
    #print(x_new)
    if np.linalg.norm(x_new-x)<eps:
        break
    x=x_new
    k+=1
    #print(k)

print('кількість ітерацій:', k)
print('розв’язок:')
print(x_new)

print('похибка:')
print(np.linalg.norm(x_new - xT))

condA=np.linalg.cond(A,2)
nu_iter=2/condA
print('теоретичний показник збіжності простої ітерації:', nu_iter)
nu_fact=(1/k)*np.log(np.linalg.norm(x-xT)/np.linalg.norm(x_new-xT))
print('фактичний показник збіжності:',nu_fact)
```

```python
#Метод Зейделя
import numpy as np

#n=3
#A=np.random.rand(n,n)
#b=np.random.rand(n)

A=np.array([[10,2,1],[2,8,1],[1,1,5]],dtype=float)
xT=np.array([[2],[5],[8]])
b=np.dot(A,xT)

n=len(b)

x=np.zeros((n,1))
y=np.zeros((n,1))

print('A \n',A)
print('xT \n',xT)
print('b \n',b)
print(np.linalg.cond(A, 2), "число обумовленості програмно")
print(np.linalg.det(A), "визначник матриці А")
print(np.linalg.eigvals(A),"власні значення")
eigvals = np.linalg.eigvals(A)

m = np.min(eigvals)
M = np.max(eigvals)
print('m =', m)
print('M =', M)

print()
alp=2/(m+M)
print(alp,'оптимальний параметр alpha')

n=3
x=np.zeros((n,1))
k=0
eps=1e-3
while True:
    x_new = x.copy()
    for i in range(n):
        s1=0
        s2=0
        for j in range(i):
            s1+=A[i,j]*x_new[j]
        for j in range(i+1,n):
            s2+=A[i,j]*x[j]
        x_new[i]=(b[i]-s1-s2)/A[i,i]
    if np.linalg.norm(x_new-x)<eps:
        break
    x=x_new
    k+=1

print('кількість ітерацій:', k)
print('розв’язок:')
print(x_new)

print('похибка:')
print(np.linalg.norm(x_new-xT))

nu_fact=(1/k)*np.log(np.linalg.norm(x-xT)/np.linalg.norm(x_new-xT))
print('фактичний показник збіжності:',nu_fact)
```
