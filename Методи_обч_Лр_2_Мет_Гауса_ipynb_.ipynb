```python
import numpy as np
#Боцу Алина Метод Гауса

#n=3
#A=np.random.rand(n,n)
#b=np.random.rand(n)

A=np.array([[2,1,0],[1,3,1],[6,1,1]],dtype=float)
b=np.array([[2],[2],[7]],dtype=float)

print("Матрица А \n", A)
print("Векор b \n",b)
print()

def metod_gausa(A, b):
    n = len(A)
    x = np.zeros(n)
    A = A.astype(float)
    b = b.astype(float)

    for k in range(n):
        if A[k][k]==0:
            print("Error")
        else:
            for i in range(k+1,n):
                N =  A[i][k] / A[k][k]
                A[i, k:] = A[i, k:] - N * A[k, k:]
                b[i] = b[i] - N * b[k]

    for k in range(n - 1, -1, -1):
        x[k] = b[k]
        for i in range(k + 1, n):
            x[k] -= A[k][i] * x[i]
        x[k] /= A[k][k]
    return x

x = metod_gausa(A, b)
print("\nПеревірка A @ x:", x)

print(np.dot(A, x))
```
