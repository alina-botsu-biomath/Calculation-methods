```python
import numpy as np
import math
from math import *
import matplotlib.pyplot as plt
import sympy as sp

def f(x):
    return np.log(x)+x
    #return x**16

a=1
b=2
xi=np.linspace(a,b,5)
yi=f(xi)
print("xi:",xi, yi)

x=sp.Symbol("x")
print("x:",x)
n=len(xi)
print("n:",n)

L4=0
for m in range (n-1):
  pm=1
  for j in range(n-1):
    if m!=j:
      pm*=(x-xi[j])/(xi[m]-xi[j])
  L4+=yi[m]*pm
print("L4=",L4)

L4_simple=sp.expand(L4)#упрощенній
print("L4_simple:",L4_simple)
print("x==0,5",L4_simple.subs(x,0.5))

xx = np.linspace(a, b, 50)
yy_real = f(xx)
yy_interp = [L4_simple.subs(x, val) for val in xx]

plt.figure(figsize=(12,9))
plt.plot(xx, yy_real, label="Функція y = ln(x)+x", color="blue", linewidth=2)
plt.plot(xx, yy_interp, "--", label="Інтерполяційний поліном Лагранжа", color="red", linewidth=2)
plt.scatter(xi, yi, color="black", s=50, zorder=5, label="Точки інтерполяції")
plt.title("Інтерполяція функції ln(x)+x на відрізку [1, 2]", fontsize=14)
plt.xlabel("x", fontsize=12)
plt.ylabel("y", fontsize=12)
plt.legend(fontsize=11)
plt.grid(True)
plt.show()
#дз посчитать похибку

print("L4 x==1.4",L4_simple.subs(x,1.4))
print("f(1.4)", f(1.4))
print("L4(1.4)-f(1.4)",abs(L4_simple.subs(x,1.4)-f(1.4)))

#M5=max y v po x є [1,2]
#f0=np.log(x)+x
f1=(x+1)/x
df5=sp.diff(f1,x,4)
print(df5)
print(sp.expand(df5))
"""
l1 = np.linspace(-10, 10, 50)
g1=24/l1**5
plt.plot(l1,g1)
plt.title("Функція y = 24/x**5", fontsize=14)
"""
l = np.linspace(1, 2, 50)
g=24/l**5
plt.plot(l,g)
plt.title("Функція y = 24/x**5", fontsize=14)
plt.show()
print("max df5=",max(g))

print("   ")
w5=1
for i in range(n):
  p=0
  p=p+(x-xi[i])
  w5=w5*p
  print(w5,'    ',p)
print("w5=",w5)
print("w5(1.4)=",abs(w5.subs(x,1.4)))

#тут все правильно с н

print("L4(1.4)-f(1.4)=",abs(L4_simple.subs(x,1.4)-f(1.4)))
print("R4<=",max(g)/120*abs(w5.subs(x,1.4)))
print("   ")

# По Ньютону
print("По Ньютону")

xi=np.linspace(a,b,5)
yi=f(xi)
print("xi=",xi)
print("yi=",yi)

n=5
matrix=np.zeros((n,n))
v1=xi
matrix[0]=v1
print(matrix.T)
matrixT=matrix.T

for k in range(1,n):
  for i in range(0,n-k):
    matrixT[i][k]=(matrixT[i+1][k-1]-matrixT[i][k-1])/(yi[i+k]-yi[i])
print(matrixT)
r_matrix = np.round(matrixT, 3)
print("rounded_matrix для понимания как она выгядит, не работы с ней \n", r_matrix)
print("   ")

#строим Інтерполяційним многочленом Ньютона з поділеними різницями
w5=1
LN4r=r_matrix[0][1]

for i in range(n-2):
  p=0
  p=p+(x-xi[i])
  w5=w5*p
  w=w5*r_matrix[0][i+2]
  LN4r=LN4r+w
  #print(p,"    ",r_matrix[0][i],"   ",w,"   ",sum)

print("Інтерполяційний многочлен Ньютона з округленням LN4r=",LN4r)
#print(LN4.subs(x,1.25))

w5=1
LN4=matrixT[0][1]

for i in range(n-2):
  p=0
  p=p+(x-xi[i])
  w5=w5*p
  w=w5*matrixT[0][i+2]
  LN4=LN4+w
  #print(p,"    ",matrixT[0][i],"   ",w,"   ",sum)

print("Інтерполяційний многочлен Ньютона LN4=",LN4)

print(LN4.subs(x,1))
print(LN4.subs(x,1.25))
print(LN4.subs(x,1.5))
print(LN4.subs(x,1.75))
print(LN4.subs(x,2))

LN4y_xi= [LN4.subs(x, val) for val in xi]
print("без округлення",LN4y_xi)

LN4ry_xi= [LN4r.subs(x, val) for val in xi]
print("з округленням",LN4ry_xi)

plt.figure(figsize=(12,9))
LN4y= [LN4.subs(x, val) for val in xx]
LN4yr= [LN4r.subs(x, val) for val in xx]
plt.plot(xx, LN4y, "-", label="поліном Ньютона без округлення", color="green", linewidth=2)
plt.plot(xx, LN4yr, "-", label="поліном Ньютона з округленням", color="orange", linewidth=2)
plt.title("Інтерполяційний многочлен Ньютона", fontsize=14)
print("   ")
plt.figure(figsize=(12,9))
plt.plot(xx, yy_real, label="Функція y = ln(x)+x", color="blue", linewidth=2)
plt.plot(xx, LN4y, "-", label="Інтерполяційний поліном Ньютона", color="green", linewidth=2)
plt.plot(xx, yy_interp, "--", label="Інтерполяційний поліном Лагранжа", color="red", linewidth=2)
plt.scatter(xi, yi, color="black", s=50, zorder=5, label="Точки інтерполяції")
plt.title("Інтерполяція функції ln(x)+x на відрізку [1, 2]", fontsize=14)
plt.xlabel("x", fontsize=12)
plt.ylabel("y", fontsize=12)
plt.legend(fontsize=11)
plt.grid(True)
plt.show()
```
