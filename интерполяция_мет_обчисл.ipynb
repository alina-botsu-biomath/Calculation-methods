```python
#y=lnx+x

import numpy as np
import math
from math import *
import matplotlib.pyplot as plt
import sympy as sp

def f(x):
    return np.log(x)+x
    #return x**16

a=1
b=2
x=np.linspace(a,b,5)
y=f(x)
print("xi:",x, y)

k=sp.Symbol("K")
l=sp.Symbol("L")
Px=k*x+l
print(Px)

X_sum=sum(x)
Y_sum=sum(y)
XY_sum=sum(x*y)
X2_sum=sum(x**2)

print()
eq=k*X2_sum+l*X_sum
print(eq)

eq1=sp.Eq(k*X2_sum+l*X_sum,XY_sum)
print(eq1)
eq2=sp.Eq(k*X_sum+l*len(x),Y_sum)
print(eq2)

sol=sp.solve((eq1,eq2),(k,l))
print(sol)
print(sol[k])
print(sol[l])

def Ps(k,l):
  return k*x+l
P1=Ps(sol[k],sol[l])

print("P1",P1)

xi=np.linspace(a,b,5)
plt.figure(figsize=(12,9))
plt.plot(xi,P1)
plt.scatter(x,y)
plt.show()

delta1=(sum(P1-y)**2)**0.5
print("delta", delta1)

print()

#для квадрата

X_sum=sum(x)
Y_sum=sum(y)
XY_sum=sum(x*y)
X2_sum=sum(x**2)

X3_sum=sum(x**3)
X4_sum=sum(x**4)
YX2_sum=sum(y*(x)**2)

s=sp.Symbol("S")
v=sp.Symbol("V")
t=sp.Symbol("T")
Psvt=s*x**2+v*x+t
print(Psvt)

eq21=sp.Eq(s*X4_sum + v*X3_sum + t*X2_sum,YX2_sum)
print(eq21)
eq22=sp.Eq(s*X3_sum + v*X2_sum + t*X_sum,XY_sum)
print(eq22)
eq23=sp.Eq(s*X2_sum + v*X_sum + t*len(x),Y_sum)
print(eq23)

solut=sp.solve((eq21,eq22,eq23),(s,v,t))
print(solut)
print(solut[s])
print(solut[v])
print(solut[t])


def Pv(s,v,t):
  return s*x**2+v*x+t
P2=Pv(solut[s],solut[v],solut[t])
print("P2",P2)

xi=np.linspace(a,b,5)
plt.figure(figsize=(12,9))
plt.plot(xi,P2)
plt.scatter(x,y)
plt.show()

delta2=(sum(P2-y)**2)**0.5
print("delta", delta2)






```
